```python
import pandas as pd
import numpy as np
import os
import glob

# Folder containing CSV files
folder_path = "annotation_files/"

# Get all CSV file paths
csv_files = glob.glob(os.path.join(folder_path, "*.csv"))
print(csv_files)

# List to store processed DataFrames
all_dfs = []

for file in csv_files:
    # Load CSV
    df = pd.read_csv(file)

    # Apply same preprocessing
    filtered_data = df[df["Done?"] == 1]
    filtered_data = filtered_data[filtered_data["Common Name"] != "Noise"]

    # Reset index
    filtered_data = filtered_data.reset_index(drop=True)

    # Optional: track source file
    filtered_data["source_file"] = os.path.basename(file)

    # Append to list
    all_dfs.append(filtered_data)

# Combine all DataFrames
combined_df = pd.concat(all_dfs, ignore_index=True)

# Confirm unique labels
print(combined_df["Common Name"].unique())

# Save combined CSV
combined_df.to_csv("annotation_files/combined_annotations.csv", index=False)

print("Combined CSV saved successfully.")


```
